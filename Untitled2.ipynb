```python
r = int(input("Enter number of rats: "))
unit = int(input("Enter food consumed by each rat: "))
n = int(input("Enter number of houses: "))
arr = list(map(int, input("Enter food in each house: ").split()))

if arr is None:
    print(-1)
else:
    food_needed = r * unit
    food = 0
    houses = 0
    for i in range(n):
        food = food + arr[i]
        houses = houses + 1

        if food >= food_needed:
            break

    if food < food_needed:
        print(0)
    else:
        print(houses)
```
