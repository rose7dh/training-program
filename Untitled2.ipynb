```python
a=int(input("Enter the list size:"))
list=[]
for i in range(a):
  b=int(input("enter the elements:"))
  list.append(b)
largest=list[0]
for i in range(1,a):
  if(list[i]>largest):
    largest=list[i]
print("Largest element=",largest)
```
